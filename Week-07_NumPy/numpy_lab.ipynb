## **Task 1: Creating Arrays of Different Dimensions**
1. Create the following arrays using `numpy.array()`:
   - A 0-D array (scalar).
   - A 1-D array with 5 integer elements.
   - A 2-D array of shape (3x3) with integers of your choice.
   - A 3-D array with dimensions (2x2x3).

2. Print each array, its `ndim` (number of dimensions), and its `shape`.


```python

import numpy as np

# 0-D array
arr0= np.array(5)

# 1-D array
arr1 = np.array([44,28,55,88,66])

# 2-D array
arr2= np.array([[17, 11, 23], [33, 41, 45], [54, 65, 88]])

# 3-D array
arr3= np.array([[[11, 12, 13], [21, 22, 23]], [[44, 55, 67], [31, 32, 33]]])

# Print each array, its `ndim` (number of dimensions), and its `shape`.

print("0-D Array:", arr0) # Print array
print("Dimensions:", arr0.ndim) # Print its ndim
print("Shape:", arr0.shape) #Print its shape
print("_" * 30)

print("1-D Array:", arr1) # Print array
print("Dimensions:", arr1.ndim) # Print its ndim
print("Shape:", arr1.shape) #Print its shape
print("_" * 30)

print("2-D Array:", arr2) # Print array
print("Dimensions:", arr2.ndim) # Print its ndim
print("Shape:", arr2.shape) #Print its shape
print("_" * 30)

print("3-D Array:", arr3) # Print array
print("Dimensions:", arr3.ndim) # Print its ndim
print("Shape:", arr3.shape) #Print its shape
print("_" * 30)
```


## **Task 2: Exploring Array Size and Memory Consumption**
For each array in Task 1, print:
- `size` → total number of elements in the array.
- `itemsize` → memory (in bytes) consumed by each element.


```python

arrays = [arr0, arr1, arr2, arr3]
for arr in arrays:
    print(f"Array:\n{arr}")
    print(f"Size: {arr.size} , Item size: {arr.itemsize} bytes")
    print(f"Total bytes consumed:{arr.size * arr.itemsize} bytes\n")
```


## **Task 3: Iterating Over Multi-Dimensional Arrays**
Use nested loops or `np.nditer()` to iterate over all elements in the 3-D array (`arr3`) created in Task 1.


```python
print("Iteration using nested loops:")
for i in arr3:  # i is each "block" (2x3 array)
    for j in i:   # j is each "row" (1x3 array)
        for k in j: # k is each element
            print(k, end=" ")

print("\n\nIteration using np.nditer():")
for val in np.nditer(arr3):
    print(val, end=" ")
```


## **Task 4: Creating Arrays Filled with Specific Values**


## **Task 6: Performance Comparison – Lists vs NumPy Arrays**
Compare the execution time of multiplying 10 million elements in a **list** vs a **NumPy array**.


```python

import time

# Using list
list_data = list(range(10000000)) #initialize for one million numbers
start = time.time()   #start time
list_result = [x * 2 for x in list_data] # multiply each element by 2
end = time.time() #end time after computation
print(f"List processing time: {end - start:.5f} seconds")

# Using NumPy array
np_data = np.arange(10000000) #initialize for one million numbers
start = time.time() #start the timer
np_result = np_data * 2 #multiply each element by 2
end = time.time() #track the time after computation
print(f"NumPy processing time: {end - start:.5f} seconds")
```
