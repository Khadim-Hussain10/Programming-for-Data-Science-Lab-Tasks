A company, TechNova Pvt. Ltd., has conducted a survey of 200 employees across its five

```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
```

***Part 1: Bar Charts (Vertical and Horizontal)***

**Data**:

Engineering: 85

Marketing: 72

HR: 65

Finance: 78

Sales: 70



Task 1.1 — Department-wise Average Productivity
1. Create a vertical bar chart showing average productivity by department.
2. Add title, x/y labels, and custom colors.
3. Display value labels on top of bars.

```python
#Given Data
departments = ["Engineering", "Marketing", "HR", "Finance", "Sales"]
average_productivity = [85, 72, 65, 78, 70]

# Vertical bar chart
plt.figure(figsize=(7,5))
bars = plt.bar(departments, average_productivity, color=['black', 'blue', 'green', 'pink', 'yellow'])

plt.title("Department-wise Average Productivity", fontsize=14, fontweight="bold")
plt.xlabel('Departments', fontsize=12, fontweight="bold", color= "red")
plt.ylabel("Average Productivity", fontsize=12, fontweight="bold", color='red')

# Add value labels on top of bars
for bar in bars:
    yval = bar.get_height()  # ← this line and the next ones must be indented
    plt.text(bar.get_x() + bar.get_width()/2, yval + 1, str(yval),
             ha='center', va='bottom', fontsize=10, fontweight='bold')

plt.show()
```

4. Create a horizontal bar chart with custom width and color scheme.

```python
plt.figure(figsize=(7,5))
bars = plt.barh(departments, average_productivity,
                color=['purple','orange','teal','brown','gray'], height=0.6)

plt.title("Department-wise Average Productivity (Horizontal)", fontsize=14, fontweight="bold")
plt.xlabel("Average Productivity", fontsize=12, fontweight="bold", color="red")
plt.ylabel("Departments", fontsize=12, fontweight="bold", color="red")

# Labels beside bars
for bar in bars:
    xval = bar.get_width()
    plt.text(xval+0.5, bar.get_y()+bar.get_height()/2, str(xval),
             va='center', fontsize=10, fontweight='bold')

plt.show()
```

**5. Comment on which visualization communicates better and why.**

**Comment:**

The *horizontal bar chart* is easier to read because it gives more space for long department names and makes it simpler to compare productivity levels across departments.



***Part 2: Histograms (Different Types)***

Task 2.1 — Distribution of Productivity Scores

Generate random productivity data for 200 employees using:
np.random.seed(42)

productivity_scores = np.random.normal(70, 10, 200)

**1. Plot a simple histogram.**

```python
np.random.seed(42)
productivity_scores = np.random.normal(70, 10, 200)

plt.figure(figsize=(8,5))
plt.hist(productivity_scores, color='skyblue', edgecolor='black')

plt.title("Distribution of Productivity Scores", fontsize=14, fontweight="bold")
plt.xlabel("Productivity Score", fontsize=12, fontweight="bold", color="red")
plt.ylabel("Number of Employees", fontsize=12, fontweight="bold", color="red")
plt.grid(True, linestyle='--', alpha=0.7)
plt.show()
```

2. Use bins = [50, 60, 70, 80, 90, 100].


```python
plt.figure(figsize=(8,5))
plt.hist(productivity_scores, bins=[50,60,70,80,90,100], color='green', edgecolor='black')

plt.title("Histogram with Custom Bins", fontsize=14, fontweight="bold", color = "black")
plt.xlabel("Productivity Range", fontsize=12, fontweight="bold", color="red")
plt.ylabel("Frequency", fontsize=12, fontweight="bold", color="red")
plt.grid(True, linestyle='--', alpha=0.7)
plt.show()
```

**3. Create overlapping histograms for two departments.**


```python
enginer_scores = np.random.normal(75,8,200)
market_scores = np.random.normal(68,9,200)

plt.figure(figsize=(8,5))
plt.hist(enginer_scores, bins=10, alpha=0.6, color='green', label='Engineering', edgecolor='black')
plt.hist(market_scores, bins=10, alpha=0.6, color='violet', label='Marketing', edgecolor='black')

plt.title("Overlapping Histograms of Two Departments", fontsize=14, fontweight="bold")
plt.xlabel("Productivity Score", fontsize=12, fontweight="bold", color="red")
plt.ylabel("Number of Employees", fontsize=12, fontweight="bold", color="red")
plt.legend()
plt.grid(True, linestyle='--', alpha=0.8)
plt.show()
```

**4. Add grid lines, titles, and legends.**


```python
plt.figure(figsize=(8,5))
plt.hist(productivity_scores, bins=10, histtype='bar', color='teal', label='bar')
plt.hist(productivity_scores, bins=10, histtype='barstacked', color='orange', alpha=0.6, label='barstacked')
plt.hist(productivity_scores, bins=10, histtype='step', color='red', linewidth=1.5, label='step')
plt.hist(productivity_scores, bins=10, histtype='stepfilled', color='plum', alpha=0.4, label='stepfilled')

plt.title("Different Histogram Types", fontsize=14, fontweight="bold")
plt.xlabel("Productivity Score", fontsize=12, fontweight="bold", color="red")
plt.ylabel("Frequency", fontsize=12, fontweight="bold", color="red")
plt.legend()
plt.grid(True, linestyle='--', alpha=0.7)
plt.show()
```

**5. Experiment with histogram types ('bar', 'barstacked', 'step', 'stepfilled').**

**Commet:**
The histograms show that most employees have productivity scores close to 70. Using different histogram types helps us display the same data in various clear and attractive ways.

***Part 3: Pie Charts (Exploring Functionalities)***

Task 3.1 — Department-wise Employee Count

Data:

Engineering: 60

Marketing: 40

HR: 25

Finance: 30

Sales: 45


1. Create a basic pie chart showing employee distribution.
2. Add startangle=90, explode for Engineering, shadow=True, and custom colors.
3. Add labels and title.
4. Experiment with different startangles (0, 45, 180).
5. Add a legend with department names and counts.
6. Display the pie chart as a donut chart using a white circle at the center.

```python
import matplotlib.pyplot as plt

sizes = [60, 40, 25, 30, 45]
departments = ["Engineering", "Marketing", "HR", "Finance", "Sales"]
colors = ['deepskyblue', 'lightgreen', 'lightcoral', 'peachpuff', 'lavender']

plt.figure(figsize=(5,5))
plt.pie(sizes, labels=departments, colors=colors, autopct='%1.1f%%')
plt.title("Employee Distribution by Department", fontsize=14, fontweight="bold")
plt.show()
```

```python
explode = [0.1,0,0,0,0]

plt.figure(figsize=(6,6))
plt.pie(sizes, labels=departments, colors=colors, autopct='%1.1f%%',
        startangle=90, explode=explode, shadow=True)
plt.title("Employee Distribution (Startangle=90)", fontsize=14, fontweight="bold")
plt.legend([f"{d} ({s})" for d,s in zip(departments,sizes)], loc='best')
plt.show()
```

```python
# Experiment with different startangles
for angle in [0, 45, 180]:
    plt.figure(figsize=(6,6))
    plt.pie(sizes, labels=departments, colors=colors, autopct='%1.1f%%',
            startangle=angle, shadow=True)
    plt.title(f"Employee Distribution (Startangle={angle})", fontsize=14, fontweight="bold")
    plt.show()
```

```python
fig, ax = plt.subplots(figsize=(7,7))
wedges, texts, autotexts = ax.pie(sizes, colors=colors, startangle=90, autopct='%1.1f%%', pctdistance=0.85)
centre_circle = plt.Circle((0,0),0.70,fc='white')
fig.gca().add_artist(centre_circle)

plt.title("Donut Chart: Department-wise Employee Distribution", fontsize=14, fontweight="bold")
plt.legend(departments, loc='best')
plt.show()
```

**Pie/Donut Chart:**

Pie charts help us see how different parts make up the whole, and donut charts make it even easier to read and look nicer.



***Part 4: Summary Challenge***

Combine results into a multi-plot figure using subplot().

In subplot 1 → Bar chart of average productivity

In subplot 2 → Histogram of productivity distribution

In subplot 3 → Pie chart of department-wise employees

Use plt.tight_layout() to adjust spacing.


```python
fig, axes = plt.subplots(1,3, figsize=(18,5))

# 1. Bar Chart
axes[0].bar(departments, average_productivity, color='skyblue', edgecolor='black')
axes[0].set_title("Average Productivity by Department", fontsize=12, fontweight="bold")
axes[0].set_xlabel("Departments")
axes[0].set_ylabel("Avg Productivity")

# 2. Histogram
axes[1].hist(productivity_scores, bins=10, color='orange', edgecolor='black')
axes[1].set_title("Productivity Distribution", fontsize=12, fontweight="bold")
axes[1].set_xlabel("Score")
axes[1].set_ylabel("Count")

# 3. Pie Chart
axes[2].pie(sizes, labels=departments, colors=colors, autopct='%1.1f%%', startangle=90)
axes[2].set_title("Employee Distribution", fontsize=12, fontweight="bold")

plt.tight_layout()
plt.show()
```

**Combine results:**

This chart combines different views to give a clear picture of the company is performance—bars show department averages, the histogram shows score distribution, and the pie chart shows staff proportions
