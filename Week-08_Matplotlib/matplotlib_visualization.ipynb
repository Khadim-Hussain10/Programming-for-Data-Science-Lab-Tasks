```python
# import required libraries
import numpy as np
import matplotlib as mp
import matplotlib.pyplot as plt

# To display plots in line
%matplotlib inline
```

>**Tasks**

**Task 1 – Basic Linear Plot**
1. Plot Study Hours vs Exam Score using plt.plot().
2. Use the fmt argument to define line color, marker, and style.
3. Try solid, dashed, dotted, dash-dot lines.
4. Change line color, width, and marker size.

```python
# DataSet
students = ["A", "B", "C", "D", "E", "F", "G", "H", "I", "J"]
Study_hours = [2.5, 3.5, 4.0, 4.5, 5.0, 5.5, 6.0, 6.5, 7.0, 8.0]
sleep_hours = [8.0, 7.5, 6.0, 5.5, 6.5, 5.0, 7.0, 6.0, 5.5, 5.0]
exam_score = [55, 60, 66, 68, 72, 74, 78, 82, 85, 90]

#convert to numpy array
study = np.array(Study_hours)
sleep = np.array(sleep_hours)
scores = np.array(exam_score)

# Task 1 – Basic Linear Plot
plt.figure(figsize=(10,6))
plt.plot(study, scores, "ro-", label="Solid", linewidth=2, markersize=8)
plt.plot(study, scores, "b--s", label="Dashed", linewidth=2, markersize=7)
plt.plot(study, scores, "g:.", label="Dotted", linewidth=3, markersize=10)
plt.plot(study, scores, "m-.^", label="Dash-dot", linewidth=2, markersize=9)

# Add labels and title
plt.title("Study Hours vs Exam Score")
plt.xlabel("Study Hours")
plt.ylabel("Exam Score")

# Show legend and grid
plt.legend()
plt.grid(True, alpha=0.2)

# Display
plt.show()
```

**Task 2 – Plot Customization**
1. Add title, x-label, and y-label.
2. Change font family, font size, and color.
3. Move the title (left, center, right).
4. Show grid lines with different colors and thickness.
5. Show grid only on x-axis or y-axis.

```python
# Task 2 – Plot Customization
plt.figure(figsize=(10, 6))

# Font dictionaries
title_font = {'family': 'serif', 'size': 16, 'color': 'darkred'}
xlabel_font = {'family': 'sans-serif', 'size': 12, 'color': 'darkgreen'}
ylabel_font = {'family': 'serif', 'size': 12, 'color': 'darkblue'}

# Basic plot with labels
plt.figure(figsize=(10,6))
plt.plot(study, scores)
plt.title('Study Hours vs Exam Score', fontdict=title_font)
plt.xlabel('Study Hours', fontdict=xlabel_font)
plt.ylabel('Exam Score', fontdict=ylabel_font)
plt.show()
print("\n" + "_--_--" * 20)

# Title at center
plt.figure(figsize=(10,6))
plt.plot(study, scores)
plt.title('Title Center', fontdict=title_font, loc='center')
plt.xlabel('Study Hours', fontdict=xlabel_font)
plt.ylabel('Exam Score', fontdict=ylabel_font)
plt.show()
print("\n" + "_--_--" * 20)

# Title at right
plt.figure(figsize=(10,6))
plt.plot(study, scores)
plt.title('Title Right', fontdict=title_font, loc='right')
plt.xlabel('Study Hours', fontdict=xlabel_font)
plt.ylabel('Exam Score', fontdict=ylabel_font)
plt.show()
print("\n" + "_--_--" * 20)

# Title at left
plt.figure(figsize=(10,6))
plt.plot(study, scores)
plt.title('Title Left', fontdict=title_font, loc='left')
plt.xlabel('Study Hours', fontdict=xlabel_font)
plt.ylabel('Exam Score', fontdict=ylabel_font)
plt.show()
print("\n" + "_--_--" * 20)

# Grid customization (major and minor)
plt.figure(figsize=(10,6))
plt.plot(study, scores)
plt.title('Grid Major and Minor', fontdict=title_font)
plt.xlabel('Study Hours', fontdict=xlabel_font)
plt.ylabel('Exam Score', fontdict=ylabel_font)
plt.grid(True, which='major', color='gray', linestyle='--', linewidth=0.8)
plt.grid(True, which='minor', axis='y', color='lightblue', linestyle=':', linewidth=0.5)
plt.minorticks_on()
plt.show()
print("\n" + "_--_--" * 20)

# Grid only on x-axis
plt.figure(figsize=(10,6))
plt.plot(study, scores)
plt.title('Grid Only X-axis', fontdict=title_font)
plt.xlabel('Study Hours', fontdict=xlabel_font)
plt.ylabel('Exam Score', fontdict=ylabel_font)
plt.grid(True, color='red', linestyle='--', linewidth=1.0, axis='x')
plt.show()
print("\n" + "_--_--" * 20)

# Grid only on y-axis
plt.figure(figsize=(10,6))
plt.plot(study, scores)
plt.title('Grid Only Y-axis', fontdict=title_font)
plt.xlabel('Study Hours', fontdict=xlabel_font)
plt.ylabel('Exam Score', fontdict=ylabel_font)
plt.grid(True, color='red', linestyle='--', linewidth=1.0, axis='y')
plt.show()
print("\n" + "_The End Task 2_" * 7)
```

**Task 3 – Marker and Line Attributes**
1. Modify marker size, face color, and edge color.
2. Use various marker shapes (circle, square, triangle, etc.).
3. Draw multiple line plots (e.g., Exam Score vs Study Hours and Sleep Hours).

```python
# Task 3 – Marker and Line Attributes
plt.figure(figsize=(10,6))

# Exam Score vs Study Hours
plt.plot(study, scores,
         marker='o', markersize=10, markerfacecolor='yellow', markeredgecolor='red', markeredgewidth=1.5,
         linestyle='-', color='blue', linewidth=2, label='Exam Score vs Study Hours')

# Exam Score vs Sleep Hours
plt.plot(sleep, scores,
         marker='s', markersize=10, markerfacecolor='green', markeredgecolor='black', markeredgewidth=1.5,
         linestyle='--', color='magenta', linewidth=2, label='Exam Score vs Sleep Hours')

# Labels and title
plt.title('Task 3: Marker and Line Attributes', fontsize=16, fontfamily='serif', color='darkblue')
plt.xlabel('Hours', fontsize=14, fontfamily='monospace', color='green')
plt.ylabel('Exam Score', fontsize=14, fontfamily='monospace', color='purple')

# Grid and legend
plt.grid(True, linestyle='-.', alpha=0.6)
plt.legend()

plt.show()
```

**Task 4 – Scatter Plots**
1. Draw a scatter plot of Study Hours vs Exam Score.
2. Draw another of Sleep Hours vs Exam Score.
3. Display both using subplots.

```python
# Task 4 – Scatter Plots
plt.figure(figsize=(10, 6))

# Scatter: Study Hours vs Exam Score
plt.subplot(1, 2, 1)  # 1 row, 2 column
plt.scatter(study, scores, color='blue', marker='o', s=80, edgecolor='black')
plt.title('Study Hours vs Exam Score', fontsize=14, fontfamily='serif', color='red')
plt.xlabel('Study Hours', fontsize=12, color='green')
plt.ylabel('Exam Score', fontsize=12, color='purple')
plt.grid(True, linestyle='--', alpha=0.6)

# Scatter: Sleep Hours vs Exam Score
plt.subplot(1, 2, 2)  # 1 row, 2 column
plt.scatter(sleep, scores, color='orange', marker='s', s=80, edgecolor='black')
plt.title('Sleep Hours vs Exam Score', fontsize=14, fontfamily='serif', color='red')
plt.xlabel('Sleep Hours', fontsize=12, color='green')
plt.ylabel('Exam Score', fontsize=12, color='purple')
plt.grid(True, linestyle='--', alpha=0.6)

plt.tight_layout()
plt.show()
```

**Task 5 – Scatter Plot Enhancements**
1. Change scatter marker color.
2. Color each data point differently.
3. Use colormap to represent Study Hours or Sleep Hours.
4. Adjust marker size and transparency (alpha).

```python
# Study Hours vs Exam Score with colormap
plt.subplot(1, 2, 1)
scatter1 = plt.scatter(study, scores,
                       c=study,            # color represents Study Hours
                       cmap='viridis',     # colormap
                       s=100,              # marker size
                       edgecolor='black',
                       alpha=0.8,          # transparency
                       marker='o')
plt.title('Study Hours vs Exam Score', fontsize=14, fontfamily='serif', color='red')
plt.xlabel('Study Hours', fontsize=12, color='green')
plt.ylabel('Exam Score', fontsize=12, color='purple')
plt.grid(True, linestyle='--', alpha=0.6)
plt.colorbar(scatter1, label='Study Hours')  # show color scale

# Sleep Hours vs Exam Score with colormap
plt.subplot(1, 2, 2)
scatter2 = plt.scatter(sleep, scores,
                       c=sleep,
                       cmap='plasma',
                       s=120,
                       edgecolor='black',
                       alpha=0.7,
                       marker='s')
plt.title('Sleep Hours vs Exam Score', fontsize=14, fontfamily='serif', color='red')
plt.xlabel('Sleep Hours', fontsize=12, color='green')
plt.ylabel('Exam Score', fontsize=12, color='purple')
plt.grid(True, linestyle='--', alpha=0.6)
plt.colorbar(scatter2, label='Sleep Hours')

plt.tight_layout()
plt.show()
```

**Task 6 – Combined Visualization**
1. Create a 2×1 subplot layout.
2. Top: Line plot (Study Hours vs Exam Score).
3. Bottom: Scatter plot (Sleep Hours vs Exam Score) with colormap.
4. Add an overall title “Visualizing Student Study Patterns”.

```python
# Task 6 – Combined Visualization
plt.figure(figsize=(10, 10))

# Top: Line plot (Study Hours vs Exam Score)
plt.subplot(2, 1, 1) # 1 Row as 2 column
plt.plot(study, scores, marker='o', linestyle='-', color='blue', linewidth=2, markersize=8,
         markerfacecolor='yellow', markeredgecolor='red', markeredgewidth=1.5)
plt.title('Study Hours vs Exam Score (Line Plot)', fontsize=14, fontfamily='serif', color='darkblue')
plt.xlabel('Study Hours', fontsize=12, color='green')
plt.ylabel('Exam Score', fontsize=12, color='purple')
plt.grid(True, linestyle='--', alpha=0.6)

# Bottom: Scatter plot (Sleep Hours vs Exam Score) with colormap
plt.subplot(2, 1, 2)  # 2 rows, 1 column, second subplot
scatter = plt.scatter(sleep, scores,
                      c=sleep,
                      cmap='plasma',
                      s=100,
                      edgecolor='black',
                      alpha=0.8,
                      marker='s')
plt.title('Sleep Hours vs Exam Score (Scatter Plot)', fontsize=14, fontfamily='serif', color='blue')
plt.xlabel('Sleep Hours', fontsize=12, color='green')
plt.ylabel('Exam Score', fontsize=12, color='purple')
plt.grid(True, linestyle='--', alpha=0.6)
plt.colorbar(scatter, label='Sleep Hours')

# Overall title
plt.suptitle('Visualizing Student Study Patterns', fontsize=16, fontfamily='serif', color='red', y=0.95)

# adjust layout
plt.tight_layout(rect=[0, 0, 1, 0.94])
# Display
plt.show()
```
