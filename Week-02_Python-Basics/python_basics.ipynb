
## Part 2: Strings in Python

1. Create a string variable `s = "DataScience"`.  
2. Access characters using positive and negative indexing.  
3. Slice substrings (first 4 chars, last 4 chars, middle part).  
4. Use stride (step size) in slicing. Example: Extract every second character.  
5. Use `len()` to find string length. Concatenate two strings.  
6. Demonstrate immutability of strings by trying to change a character.  
7. Use escape sequences like `\n`, `\t`, and quotes inside strings.  
8. Apply string methods: `.upper()`, `.lower()`, `.replace("Data", "AI")`.  


```

s = "DataScience"

# Indexing
print(s[0])     # Positive index
print(s[-1])    # Negative index

# Slicing
print(s[0:4])   # First 4 chars
print(s[-4:])   # Last 4 chars
print(s[4:8])   # Middle

# Stride
print(s[::2])   # Every 2nd char

# Length and concatenation
print(len(s))
s2 = "Lab"
print(s + s2)

# Immutability
try:
    s[0] = "X"   # This will cause an error
except TypeError as e:
    print("Error:", e)

# Escape sequences
print("Hello\nWorld")
print("Column1\tColumn2")
print("She said, \"Python is fun!\"")

# String methods
print(s.upper())
print(s.lower())
print(s.replace("Data", "AI"))
```


## Part 3: Lists and Tuples

1. Create a list `L = [1, 2, 3, "Data", [4, 5]]` and a tuple `T = (10, 20, 30, "Science")`.  
2. Slice both list and tuple.  
3. Show immutability of tuples (try changing one element).  
4. Nesting: Access the nested list element inside `L`.  
5. Concatenate two lists, and use `extend()` to add elements.  
6. Modify list values (change index 0 to `"Python"`).  
7. Use `split()` on a string `"Data,Science,Python"` and store results in a list.  
8. Demonstrate aliasing: assign `L2 = L` and modify `L2`. What happens to `L`?  
9. Demonstrate cloning: `L3 = L[:]`. Modify `L3` and compare with `L`.  


```

L = [1, 2, 3, "Data", [4, 5]]
T = (10, 20, 30, "Science")

# Slicing
print(L[0:3])
print(T[1:])

# Tuples immutability
try:
    T[0] = 99
except TypeError as e:
    print("Error:", e)

# Nesting
print(L[4][0])

# Concatenation & extend
L2 = [100, 200]
print(L + L2)
L.extend([300, 400])
print(L)

# Modifying list values
L[0] = "Python"
print(L)

# Split string
text = "Data,Science,Python"
print(text.split(","))

# Aliasing
L2 = L
L2[1] = "Aliased"
print("L:", L)
print("L2:", L2)

# Cloning
L3 = L[:]
L3[2] = "Cloned"
print("L:", L)
print("L3:", L3)
```
