### Step 1. Data Input (5 marks)
Create a text file `students.txt` with:
```
Ali,85
Sara,90
Omar,abc
Hina,78
```
Read file, skip invalid numeric entries, and store valid entries in a list of tuples.

````python
students = []

with open("students.txt", "w") as f: # Create file
    f.write("Ali,85 \n Sara,90 \n Omar,abc \n Hina,78")

with open("students.txt", "r") as file:
    for line in file:
        name, score = line.strip().split(",")
        try:
            students.append((name, int(score)))
        except ValueError:
            continue #skip invalid entries
students
````

### Step 2. Define a Class (5 marks)
Create a class `Student` with attributes, `grade()` method, class variable counter, and `__repr__()` method.

````python
class Student:
    count = 0

    def __init__(self, name, score):
        self.name = name
        self.score = score
        Student.count += 1

    def grade(self):
        return "Pass" if self.score >= 80 else "Fail"

    def __repr__(self):
        return f"{self.name}: {self.score} ({self.grade()})"

student_objects = [Student(name, score) for name, score in students]  # Create objects for valid data

print("Student Objects:")
for s in student_objects:
    print(s)
print("Total students:", Student.count)
````

### Step 3. Create and Analyze DataFrame (5 marks)
Convert valid students into DataFrame, add Grade column, and compute statistics.

````python
import pandas as pd

df = pd.DataFrame(students, columns=["Name", "Score"]) #Create DataFrame

df["Grade"] = df["Score"].apply(lambda s: "Pass" if s >= 80 else "Fail") #Grade column

#Display
print("Average Score:", df["Score"].mean())
print("\nGrade Distribution:")
print(df["Grade"].value_counts())
print("\nDataFrame:")
print(df)
````

### Step 4. Functional Analysis (5 marks)
Write function `top_students(df, threshold)`.

````python
def top_students(df, threshold):
    return df[df["Score"] > threshold]

print("\nTop Students (Score > 80):")
print(top_students(df, 80))
````

### Step 5. Export Results (5 marks)
Export final DataFrame to `results.csv`.

````python
df.to_csv("results.csv", index=False)
print("Results exported to results.csv")
````
