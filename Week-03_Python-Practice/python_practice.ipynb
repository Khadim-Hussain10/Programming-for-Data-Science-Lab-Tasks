You are given student data:
```python
students = {

}
```
1. Add a new student `"Eve": 78`.
2. Update `"Bob"`’s marks to `80`.
3. Print the names of all students who scored **>= 75**.
4. Find and print the **student with the highest marks**.

````python
from re import T
students = {
    "Alice": 85,
    "Bob": 72,
    "Charlie": 90,
    "Diana": 65
}

# Your code here

#Add new student Eve 78 marks
students["Eve"] = 78
print(students)

#Update "Bob" 's marks to 80
students["Bob"] = 80
print(students)

#Print the names of all students who scored >= 75.
print("Students with marks >= 75:")
for name, marks in students.items():
    if marks >= 75:
        print(name)

#Find and print the student with the highest marks.
highest_student = None
highest_marks = -1

for name, marks in students.items():
  if marks > highest_marks:
    highest_student = name
    highest_marks = marks
    print(highest_student, highest_marks)

print("Topper of the student is:", highest_student, "with", students[highest_student], "marks")
````

## **Task 5 – Functions (15 minutes)**
Define a function `is_prime(n)` that:
- Returns `True` if `n` is a prime number, otherwise `False`.

Then, write code that:
1. Calls the function for numbers from 1 to 20.
2. Prints all prime numbers in that range.

````python
# Your code here

# Define a function to check if a number is prime
def is_prime(n):
  if n < 2:
    return False

  for i in range(2, n):
    if n % i == 0:
      return False
  return True

print("Print prime numbers from 1 to 20:")

for num in range(1, 21):
  if is_prime(num):
    print(num)
````
