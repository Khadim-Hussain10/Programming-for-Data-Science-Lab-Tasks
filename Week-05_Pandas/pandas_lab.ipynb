4. Rename columns to replace spaces with underscores for easier access (e.g.,

```python
df.columns = df.columns.str.replace(' ', '_').str.lower()
print("Renamed columns:")
print(df.columns.tolist())
```

# **Part 2 — Data Cleaning and Transformation**

5. Check for and handle missing or duplicate rows. Drop duplicates if any, and handle
missing values appropriately (justify your choice).


```python
# Check for duplicates
print("Number of duplicate rows before dropping:", df.duplicated().sum())
```

```python
# Drop duplicates
df = df.drop_duplicates()
print("Number of duplicate rows after dropping:", df.duplicated().sum())
```

```python
# Check for missing values
print("Missing values before handling:")
print(df.isnull().sum())
```

```python
# Handle missing values (if any)
# For this dataset, usually there are none, but if any exist:
# Then for Numeric columns: fill with mean
# For Object type or Categorical columns: fill with mode
for col in df.columns:
    if df[col].dtype != 'object':
        df[col] = df[col].fillna(df[col].mean())
    else:
        df[col] = df[col].fillna(df[col].mode()[0])

print("Missing values after handling:")
print(df.isnull().sum())
```

6. Create a new column average_score as the mean of (math score, reading score, writing
score).


```python
df['average_score'] = df[['math_score', 'reading_score', 'writing_score']].mean(axis=1)
```

```python
df
```

7. Categorize students based on average_score into Fail (0–59), Pass (60–69), Merit (70–
79), and Distinction (80–100). Store this in a new column performance_category.


```python
# Step 7: Categorize students based on average_score
def categorize(score):
    if score < 60:
        return 'Fail'
    elif score < 70:
        return 'Pass'
    elif score < 80:
        return 'Merit'
    else:
        return 'Distinction'

df['performance_category'] = df['average_score'].apply(categorize)
df
```

8. Convert gender, parental level of education, and test_prep columns to categorical data
types using astype('category').

```python
# Convert columns to categorical data type
df[['gender', 'parental_level_of_education', 'test_preparation_course']] = df[['gender', 'parental_level_of_education', 'test_preparation_course']].astype('category')
```

```python
df.dtypes
```

# **Part 3 — Data Analysis and Aggregation**

9. Compute the average math, reading, and writing scores by gender.

```python
# Compute the average math, reading, and writing scores by gender.
avg_scores_by_gender = df.groupby('gender', observed=True)[['math_score', 'reading_score', 'writing_score']].mean()
print("🔹 Average scores by gender:")
display(avg_scores_by_gender)
```

10. Find average scores grouped by: (a) Gender and test_prep, (b) Parental level of
education.

```python
# (a) Average scores grouped by gender and test preparation course
avg_by_gender_prep = df.groupby(['gender', 'test_preparation_course'], observed = True)[['math_score', 'reading_score', 'writing_score', 'average_score']].mean()
print("Average scores by Gender and Test Preparation Course:")
display(avg_by_gender_prep)
```

```python
# (b) Average scores grouped by parental level of education
avg_by_parent_edu = df.groupby('parental_level_of_education', observed = True)[['math_score', 'reading_score', 'writing_score', 'average_score']].mean()
print("Average scores by Parental Level of Education:")
display(avg_by_parent_edu)
```

11. Identify which test preparation group performed best on average (based on
average_score).


```python
# Step 11
best_test_prep = df.groupby('test_preparation_course', observed=True)['average_score'].mean().sort_values(ascending=False)
print("Average performance by Test Preparation Course:")
display(best_test_prep)

best_group = best_test_prep.idxmax()
print(f"The best performing group is: {best_group}")
```

12. Sort the dataset by average_score in descending order and display the top 10
performers.


```python
# Step 12
top_10 = df.sort_values(by='average_score', ascending=False).head(10)
print("Top 10 performers based on Average Score:")
display(top_10)
```

13. Calculate and print: (a) Number of students in each performance_category, (b)
Percentage of students who completed the test preparation course.

```python
# (a) Number of students in each performance category
performance_counts = df['performance_category'].value_counts()
print("🔹 Number of students in each performance category:")
display(performance_counts)
```

```python
# (b) Percentage of students who completed the test preparation course
completed_percentage = (df['test_preparation_course'].value_counts(normalize=True) * 100).get('completed', 0)
print(f"Percentage of students who completed the test preparation course: {completed_percentage:.2f}%")
```
